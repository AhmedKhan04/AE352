```python
import scipy.integrate as sp 
import numpy as np 

g = 9.81
m = 500
x_0, z_0, vx_0, vz_0 = 0, 1000, 1000, 0 
state_0 = [x_0, z_0, vx_0, vz_0]
def ODE(time, state): 
    v = np.sqrt(state[2]**2 + state[3]**2)
    dvx = -0.14 * v * state[2]/m
    dvz = -g - (0.14 * v * state[3]/m)
    dx = state[2]
    dz = state[3]

    return np.array([dx, dz, dvx, dvz])

solution = sp.solve_ivp(ODE,(0,10000), state_0, method='RK45', rtol = 1e-12, atol = 1e-12)
solution_states  = solution.y
z_positions = solution_states[1]

index_hit_ground = np.where(z_positions <= 1)[0][0]

time_ground = solution.t[index_hit_ground]
x_pos_ground = solution.y[0][index_hit_ground]

print(f"time to hit ground: {time_ground} seconds")
print(f"x_cord to hit ground: {x_pos_ground} meters")
```
